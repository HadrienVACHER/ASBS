```python
import wandb
import pandas as pd

api = wandb.Api()
run = api.run("CCDS_NTU/adjoint_sampler/runs/o26lu52a")

# 1. Récupérer toutes les métriques ligne par ligne (itérateur)
history_data = run.scan_history()

# 2. Convertir en DataFrame Pandas
df = pd.DataFrame(history_data)

# 3. Sauvegarder dans un fichier CSV
df.to_csv("run_o26lu52a_history.csv", index=False)

print(f"Historique sauvegardé : {len(df)} lignes exportées.")
```
